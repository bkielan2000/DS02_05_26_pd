Zadanie 7  Długości słów
Masz listę słów. Używając list comprehension, stwórz nową listę zawierającą długości tych słów.
Dataset:
slowa = ["python", "data", "science", "machine", "learning", "AI"]
Wymagania:
Użyj list comprehension: [len(slowo) for slowo in slowa]
Dodatkowo: dict comprehension słowo → długość
Oczekiwany rezultat:
Lista długości: [6, 4, 7, 7, 8, 2]
Słownik: {"python": 6, "data": 4, ...}

```python
slowa = ["python", "data", "science", "machine", "learning", "AI"]
dl_slowa = [len(slowo) for slowo in slowa] 
print("Lista długości:", dl_slowa)
slownik = {slowo: dl_slowo for slowo, dl_slowo in zip(slowa, dl_slowa)}
print(slownik)
```

Zadanie 8  Map i filter
Używając map() i filter() z lambda, przetwórz listę liczb: najpierw zostaw tylko dodatnie, potem
podwój każdą.
Dataset:
liczby = [-5, 3, -2, 8, 0, -1, 7, 4, -9, 6]

filter() z lambda: x > 0
map() z lambda: x * 2
Konwertuj wynik na listę
Oczekiwany rezultat:

```python
liczby = [-5, 3, -2, 8, 0, -1, 7, 4, -9, 6]
dodatnie = list(filter(lambda liczba: liczba>0, liczby))
podwojone = list(map(lambda dodatnia: dodatnia*2, dodatnie))
podwojone
```

Zadanie 13  Pipeline przetwarzania danych sprzedażowych
Stwórz kompletny pipeline do analizy danych sprzedażowych. Dane zawierają: dzień, kategoria, liczba
transakcji, wartość.
Dataset:
sprzedaz = [
(1, "elektronika", 45, 12500),
(1, "odziez", 78, 4200),
(2, "elektronika", 52, 15200),
(2, "odziez", 65, 3800),
(3, "elektronika", 38, 9800),
(3, "odziez", 82, 5100),
(4, "elektronika", 61, 18900),
(4, "odziez", 71, 4500),
Wymagania:
Funkcja tuple_to_dict() - konwersja tuple na dict
Funkcja dodaj_srednia_transakcji() - feature engineering (wartosc / liczba_transakcji)
Funkcja agreguj_po_kategorii() - suma wartości per kategoria (użyj dict.get() )
Użyj comprehensions do przetwarzania
Sortowanie kategorii według wartości (malejąco) przez sorted() + lambda
Oczekiwany rezultat:
Ranking kategorii z sumą wartości sprzedaży

```python
sprzedaz = [ (1, "elektronika", 45, 12500), (1, "odziez", 78, 4200), (2, "elektronika", 52, 15200), (2, "odziez", 65, 3800), (3, "elektronika", 38, 9800), (3, "odziez", 82, 5100), (4, "elektronika", 61, 18900), (4, "odziez", 71, 4500)]

def tuple_to_dict(dane):
    return [
        {
            "dzien": dzien,
            "kategoria": kategoria,
            "liczba_transakcji": liczba_transakcji,
            "wartosc": wartosc
        }
        for dzien, kategoria, liczba_transakcji, wartosc in dane
    ]

def dodaj_srednia_transakcji(dane):
    return [
        {
            **rekord,
            "srednia_transakcja": round(rekord["wartosc"] / rekord["liczba_transakcji"],2)
        }
        for rekord in dane
    ]

def agreguj_po_kategorii(dane):
    wynik = {}

    for rekord in dane:
        kategoria = rekord["kategoria"]
        wynik[kategoria] = wynik.get(kategoria, 0) + rekord["wartosc"]

    return wynik

dane_dict = tuple_to_dict(sprzedaz)
dane_srednia = dodaj_srednia_transakcji(dane_dict)
agregacja = agreguj_po_kategorii(dane_srednia)

ranking = sorted(
    agregacja.items(),
    key=lambda x: x[1],
    reverse=True
)


print("Suma wartości sprzedaży wg kategorii:")
print(agregacja)

print("Ranking kategorii:")
print(ranking)
```
